```python
# ============================================================
# LOAN APPROVAL & RISK ANALYTICS
# File: 03_eda_business_analysis.ipynb
# Purpose: Exploratory Data Analysis & Business Insights
# ============================================================


# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)

sns.set_theme(style="whitegrid")

print("Libraries imported successfully!")


# ============================================================
# 2. LOAD CLEANED DATASET
# ============================================================

DATA_PATH = "../data/loan_cleaned.csv"

df = pd.read_csv(DATA_PATH)

print("Cleaned dataset loaded successfully!")

print(f"Rows    : {df.shape[0]}")
print(f"Columns : {df.shape[1]}")


# ============================================================
# 3. BASIC DATASET OVERVIEW
# ============================================================

print("\n" + "=" * 70)
print("DATASET OVERVIEW")
print("=" * 70)

print(df.head())

print("\nData Types:")
print(df.dtypes)


# ============================================================
# 4. LOAN APPROVAL OVERVIEW
# ============================================================

print("\n" + "=" * 70)
print("LOAN APPROVAL OVERVIEW")
print("=" * 70)

loan_status = df["Loan_Status"].value_counts()

print(loan_status)

approval_rate = (
    (df["Loan_Status"] == "Y").mean() * 100
)

rejection_rate = (
    (df["Loan_Status"] == "N").mean() * 100
)

print(f"\nApproval Rate : {approval_rate:.2f}%")
print(f"Rejection Rate: {rejection_rate:.2f}%")


# ============================================================
# 5. LOAN STATUS DISTRIBUTION
# ============================================================

plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="Loan_Status"
)

plt.title("Loan Approval Distribution")
plt.xlabel("Loan Status")
plt.ylabel("Number of Applications")

plt.tight_layout()
plt.show()


# ============================================================
# 6. CREDIT HISTORY ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("CREDIT HISTORY ANALYSIS")
print("=" * 70)

credit_analysis = pd.crosstab(
    df["Credit_History"],
    df["Loan_Status"],
    normalize="index"
) * 100

print(
    credit_analysis.round(2)
)


# Credit History vs Loan Status

plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Credit_History",
    hue="Loan_Status"
)

plt.title("Credit History vs Loan Approval")
plt.xlabel("Credit History")
plt.ylabel("Number of Applicants")

plt.tight_layout()
plt.show()


# Credit History approval rate

credit_approval_rate = (
    df.groupby("Credit_History")["Loan_Status"]
    .apply(lambda x: (x == "Y").mean() * 100)
)

print("\nApproval Rate by Credit History:")
print(credit_approval_rate.round(2))


# ============================================================
# 7. APPLICANT INCOME ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("APPLICANT INCOME ANALYSIS")
print("=" * 70)

print(
    df.groupby("Loan_Status")["ApplicantIncome"]
    .agg(["mean", "median", "min", "max"])
    .round(2)
)


plt.figure(figsize=(9, 5))

sns.boxplot(
    data=df,
    x="Loan_Status",
    y="ApplicantIncome"
)

plt.title("Applicant Income vs Loan Approval")
plt.xlabel("Loan Status")
plt.ylabel("Applicant Income")

plt.tight_layout()
plt.show()


# ============================================================
# 8. CO-APPLICANT INCOME ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("CO-APPLICANT INCOME ANALYSIS")
print("=" * 70)

print(
    df.groupby("Loan_Status")["CoapplicantIncome"]
    .agg(["mean", "median", "min", "max"])
    .round(2)
)


plt.figure(figsize=(9, 5))

sns.boxplot(
    data=df,
    x="Loan_Status",
    y="CoapplicantIncome"
)

plt.title("Co-applicant Income vs Loan Approval")
plt.xlabel("Loan Status")
plt.ylabel("Co-applicant Income")

plt.tight_layout()
plt.show()


# ============================================================
# 9. TOTAL HOUSEHOLD INCOME
# ============================================================

df["TotalIncome"] = (
    df["ApplicantIncome"] +
    df["CoapplicantIncome"]
)

print("\n" + "=" * 70)
print("TOTAL HOUSEHOLD INCOME ANALYSIS")
print("=" * 70)

print(
    df.groupby("Loan_Status")["TotalIncome"]
    .agg(["mean", "median", "min", "max"])
    .round(2)
)


plt.figure(figsize=(9, 5))

sns.boxplot(
    data=df,
    x="Loan_Status",
    y="TotalIncome"
)

plt.title("Total Household Income vs Loan Approval")
plt.xlabel("Loan Status")
plt.ylabel("Total Income")

plt.tight_layout()
plt.show()


# ============================================================
# 10. LOAN AMOUNT ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("LOAN AMOUNT ANALYSIS")
print("=" * 70)

print(
    df.groupby("Loan_Status")["LoanAmount"]
    .agg(["mean", "median", "min", "max"])
    .round(2)
)


plt.figure(figsize=(9, 5))

sns.boxplot(
    data=df,
    x="Loan_Status",
    y="LoanAmount"
)

plt.title("Loan Amount vs Loan Approval")
plt.xlabel("Loan Status")
plt.ylabel("Loan Amount")

plt.tight_layout()
plt.show()


# ============================================================
# 11. LOAN AMOUNT / INCOME RATIO
# ============================================================

df["LoanIncomeRatio"] = (
    df["LoanAmount"] /
    df["TotalIncome"]
)

print("\n" + "=" * 70)
print("LOAN TO INCOME RATIO")
print("=" * 70)

print(
    df.groupby("Loan_Status")["LoanIncomeRatio"]
    .agg(["mean", "median", "min", "max"])
    .round(4)
)


plt.figure(figsize=(9, 5))

sns.boxplot(
    data=df,
    x="Loan_Status",
    y="LoanIncomeRatio"
)

plt.title("Loan-to-Income Ratio vs Loan Approval")
plt.xlabel("Loan Status")
plt.ylabel("Loan Amount / Total Income")

plt.tight_layout()
plt.show()


# ============================================================
# 12. EDUCATION ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("EDUCATION ANALYSIS")
print("=" * 70)

education_analysis = pd.crosstab(
    df["Education"],
    df["Loan_Status"],
    normalize="index"
) * 100

print(
    education_analysis.round(2)
)


plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Education",
    hue="Loan_Status"
)

plt.title("Education vs Loan Approval")
plt.xlabel("Education")
plt.ylabel("Number of Applicants")

plt.tight_layout()
plt.show()


# ============================================================
# 13. PROPERTY AREA ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("PROPERTY AREA ANALYSIS")
print("=" * 70)

property_analysis = pd.crosstab(
    df["Property_Area"],
    df["Loan_Status"],
    normalize="index"
) * 100

print(
    property_analysis.round(2)
)


plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Property_Area",
    hue="Loan_Status"
)

plt.title("Property Area vs Loan Approval")
plt.xlabel("Property Area")
plt.ylabel("Number of Applicants")

plt.tight_layout()
plt.show()


# ============================================================
# 14. SELF EMPLOYMENT ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("SELF EMPLOYMENT ANALYSIS")
print("=" * 70)

self_employed_analysis = pd.crosstab(
    df["Self_Employed"],
    df["Loan_Status"],
    normalize="index"
) * 100

print(
    self_employed_analysis.round(2)
)


plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Self_Employed",
    hue="Loan_Status"
)

plt.title("Self Employment vs Loan Approval")
plt.xlabel("Self Employed")
plt.ylabel("Number of Applicants")

plt.tight_layout()
plt.show()


# ============================================================
# 15. MARITAL STATUS ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("MARITAL STATUS ANALYSIS")
print("=" * 70)

married_analysis = pd.crosstab(
    df["Married"],
    df["Loan_Status"],
    normalize="index"
) * 100

print(
    married_analysis.round(2)
)


plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Married",
    hue="Loan_Status"
)

plt.title("Marital Status vs Loan Approval")
plt.xlabel("Married")
plt.ylabel("Number of Applicants")

plt.tight_layout()
plt.show()


# ============================================================
# 16. DEPENDENTS ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("DEPENDENTS ANALYSIS")
print("=" * 70)

dependents_analysis = pd.crosstab(
    df["Dependents"],
    df["Loan_Status"],
    normalize="index"
) * 100

print(
    dependents_analysis.round(2)
)


plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Dependents",
    hue="Loan_Status"
)

plt.title("Dependents vs Loan Approval")
plt.xlabel("Number of Dependents")
plt.ylabel("Number of Applicants")

plt.tight_layout()
plt.show()


# ============================================================
# 17. GENDER ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("GENDER ANALYSIS")
print("=" * 70)

gender_analysis = pd.crosstab(
    df["Gender"],
    df["Loan_Status"],
    normalize="index"
) * 100

print(
    gender_analysis.round(2)
)


plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Gender",
    hue="Loan_Status"
)

plt.title("Gender vs Loan Approval")
plt.xlabel("Gender")
plt.ylabel("Number of Applicants")

plt.tight_layout()
plt.show()


# ============================================================
# 18. LOAN TERM ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("LOAN TERM ANALYSIS")
print("=" * 70)

loan_term_analysis = pd.crosstab(
    df["Loan_Amount_Term"],
    df["Loan_Status"],
    normalize="index"
) * 100

print(
    loan_term_analysis.round(2)
)


plt.figure(figsize=(10, 5))

sns.countplot(
    data=df,
    x="Loan_Amount_Term",
    hue="Loan_Status"
)

plt.title("Loan Term vs Loan Approval")
plt.xlabel("Loan Term")
plt.ylabel("Number of Applicants")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()


# ============================================================
# 19. INCOME DISTRIBUTION
# ============================================================

plt.figure(figsize=(10, 5))

sns.histplot(
    data=df,
    x="TotalIncome",
    hue="Loan_Status",
    kde=True,
    bins=30
)

plt.title("Total Household Income Distribution")
plt.xlabel("Total Household Income")
plt.ylabel("Number of Applicants")

plt.tight_layout()
plt.show()


# ============================================================
# 20. LOAN AMOUNT DISTRIBUTION
# ============================================================

plt.figure(figsize=(10, 5))

sns.histplot(
    data=df,
    x="LoanAmount",
    hue="Loan_Status",
    kde=True,
    bins=30
)

plt.title("Loan Amount Distribution")
plt.xlabel("Loan Amount")
plt.ylabel("Number of Applicants")

plt.tight_layout()
plt.show()


# ============================================================
# 21. CORRELATION ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("NUMERICAL CORRELATION")
print("=" * 70)

numeric_df = df.select_dtypes(
    include=np.number
)

correlation_matrix = numeric_df.corr()

print(
    correlation_matrix.round(2)
)


plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Numerical Feature Correlation Matrix")

plt.tight_layout()
plt.show()


# ============================================================
# 22. BUSINESS KPI SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("BUSINESS KPI SUMMARY")
print("=" * 70)

total_applications = len(df)

approved_applications = (
    df["Loan_Status"] == "Y"
).sum()

rejected_applications = (
    df["Loan_Status"] == "N"
).sum()

average_income = df["TotalIncome"].mean()

average_loan = df["LoanAmount"].mean()

print(f"Total Applications       : {total_applications}")
print(f"Approved Applications    : {approved_applications}")
print(f"Rejected Applications    : {rejected_applications}")
print(f"Approval Rate            : {approval_rate:.2f}%")
print(f"Rejection Rate           : {rejection_rate:.2f}%")
print(f"Average Household Income : {average_income:.2f}")
print(f"Average Loan Amount      : {average_loan:.2f}")


# ============================================================
# 23. CREATE BUSINESS INSIGHTS TABLE
# ============================================================

business_insights = pd.DataFrame({
    "Metric": [
        "Total Applications",
        "Approved Applications",
        "Rejected Applications",
        "Approval Rate",
        "Rejection Rate",
        "Average Household Income",
        "Average Loan Amount"
    ],
    "Value": [
        total_applications,
        approved_applications,
        rejected_applications,
        round(approval_rate, 2),
        round(rejection_rate, 2),
        round(average_income, 2),
        round(average_loan, 2)
    ]
})

print("\n" + "=" * 70)
print("BUSINESS INSIGHTS TABLE")
print("=" * 70)

print(business_insights)


# ============================================================
# 24. SAVE ANALYTICS DATASET
# ============================================================

OUTPUT_PATH = "../data/loan_eda_data.csv"

df.to_csv(
    OUTPUT_PATH,
    index=False
)

print("\nEDA dataset saved successfully!")
print(f"Location: {OUTPUT_PATH}")


# ============================================================
# 25. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("EDA & BUSINESS ANALYSIS COMPLETED")
print("=" * 70)

print("✓ Loan approval distribution analyzed")
print("✓ Credit history analyzed")
print("✓ Income analyzed")
print("✓ Loan amount analyzed")
print("✓ Loan-to-income ratio calculated")
print("✓ Education analyzed")
print("✓ Property area analyzed")
print("✓ Employment analyzed")
print("✓ Marital status analyzed")
print("✓ Dependents analyzed")
print("✓ Gender analyzed")
print("✓ Loan term analyzed")
print("✓ Correlation analysis completed")
print("✓ Business KPIs calculated")

print("\nNext Stage:")
print("Feature Engineering & ML Preparation")
```
